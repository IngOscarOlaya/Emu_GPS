```python
# pip install jupyter
# pip install PyQt5
# pip install pyserial
# pip install pyinstaller

import os
import sys
import threading

# Añadir el directorio actual al sys.path
sys.path.append(os.getcwd())

from PyQt5.QtWidgets import QApplication, QMainWindow, QSplitter
from PyQt5.QtCore import Qt
from gpswidget import GPSWidget
from admgps import admGPS

# Importar el script de envío TCP
import enviar_tcp
```

```python
class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        
        self.setWindowTitle("GPS Emulator")
        self.setGeometry(100, 100, 800, 600)
        
        splitter = QSplitter(Qt.Horizontal)
        self.gpsWidget = GPSWidget()
        self.adminWidget = admGPS()
        
        splitter.addWidget(self.gpsWidget)
        splitter.addWidget(self.adminWidget)
        
        # Establecer proporciones (60% y 40%)
        splitter.setSizes([6, 4])
        
        self.setCentralWidget(splitter)
```

```python
def run_app():
    app = QApplication.instance()  # Obtener la instancia actual de QApplication
    if app is None:
        app = QApplication(sys.argv)
    mainWin = MainWindow()
    mainWin.show()
    
    # Ejecutar enviar_tcp.py en un hilo separado
    tcp_thread = threading.Thread(target=enviar_tcp.main)
    tcp_thread.daemon = True
    tcp_thread.start()

    return app.exec_()  # No usar sys.exit() aquí

```

```python
run_app()
```
